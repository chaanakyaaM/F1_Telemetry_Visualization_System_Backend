```python
import fastf1
from fastf1.plotting import get_driver_color
from fastf1.api import driver_info

session = fastf1.get_session(2023, "Bahrain", "R")
session.load()
path = session.api_path

# Fetch driver info
drivers = driver_info(path)

# Print each driver’s headshot URL
for number, info in drivers.items():
    name = info.get('BroadcastName')
    headshot = info.get('HeadshotUrl')
    print(f"{number} — {name}: {headshot}")
```

```python
session = fastf1.get_session(2025, 22, "R")

# Load only laps and telemetry — no weather, messages, etc.
session.load(laps=True, telemetry=True)

# Pick the fastest lap
fastest_lap = session.laps.pick_fastest()

# Extract only the X/Y coordinates
track_coords = fastest_lap.get_telemetry()[["X", "Y"]]
track_coords
```
