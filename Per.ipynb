```python
import fastf1
import pandas as pd

session = fastf1.get_session(2023, "Bahrain", "R")
session.load()

lap = session.laps.pick_driver("PER")
telemetry = lap.get_telemetry()
```

```python
df = pd.DataFrame(telemetry)
df
```

```python
df.drop(columns=["DriverAhead","DistanceToDriverAhead","Status","DRS","Source"],inplace=True)
df['Time_sec'] = df['Time'].dt.total_seconds()
df['Speed'] = df['Speed'].round(2)
df["X"] = df["X"].round(3)
df["Y"] = df["Y"].round(3)
df["Z"] = df["Z"].round(3)
df['RPM'] = df["RPM"].round(3)
df['Distance'] = df["Distance"].round(3)
df.drop(columns='Time',inplace=True)
```

```python
df
```

```python
data = df[["Time_sec", "X", "Y", "Speed"]].to_dict(orient="records")
data
```

```python
import json
with open("data.js", "w") as f:
    f.write("const data = `")
    json.dump(data, f)
    f.write("`;")
    f.write("export default data;")
```
