```python
import fastf1 as f1
import time


year = 2024
round_num = 11

t1 = time.time()

session = f1.get_session(year, round_num, "R")
session.load(laps=True, telemetry=False)

schedule = f1.get_event_schedule(year)

event_name = session.event["EventName"]
event_row = schedule[schedule["EventName"] == event_name].iloc[0]
official_name = event_row["OfficialEventName"]

driver_codes = [
    session.get_driver(d)["Abbreviation"]
    for d in session.drivers
]
for code in driver_codes:
    laps = session.laps.pick_driver(code)
    print(f"{code}: {laps['Team'].iloc[0]}")
    print(f"{code}: {laps['DriverNumber'].iloc[0]}")

lap = session.laps.pick_fastest()
# coor = lap.get_telemetry()[["X", "Y"]]
try:
    coor = lap.get_telemetry()[["X", "Y"]]
except:
    print("error")
# else:
#     print("No telemetry available for this lap")

# print(event_name, official_name, driver_codes)
val = {
    "Event_name" : event_name,
    "Event_full_name" : official_name,
    "Driver_codes" : {i : session.laps.pick_driver(i)['Team'].iloc[0] for i in driver_codes},
}
print(val)
t2 = time.time()
elapsed_ms = (t2 - t1) * 1000
print(f"{elapsed_ms:.2f} ms")
```
